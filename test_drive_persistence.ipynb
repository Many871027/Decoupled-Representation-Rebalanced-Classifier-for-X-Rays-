# 🧪 Test de Persistencia de Google Drive
Este notebook verifica la conexión y la capacidad de escritura en las rutas configuradas para el proyecto de Chest X-Ray.

```python
# 1. Montaje de Google Drive
from google.colab import drive
import os
from pathlib import Path

drive.mount('/content/drive')
print("✅ Drive montado exitosamente.")
```

```python
# 2. Definición de Rutas
BASE_DRIVE_PATH = '/content/drive/MyDrive/ChestXRayProject'
MLRUNS_DIR = os.path.join(BASE_DRIVE_PATH, 'mlruns')
ARTIFACTS_DIR = os.path.join(BASE_DRIVE_PATH, 'artifacts')

print(f"Verificando ruta base: {BASE_DRIVE_PATH}")
```

```python
# 3. Creación de Carpetas y Archivos de Prueba
try:
    # Crear carpeta base si no existe
    os.makedirs(BASE_DRIVE_PATH, exist_ok=True)
    
    # Crear subcarpetas
    os.makedirs(MLRUNS_DIR, exist_ok=True)
    os.makedirs(ARTIFACTS_DIR, exist_ok=True)
    
    # Escribir archivo de prueba en mlruns
    with open(os.path.join(MLRUNS_DIR, 'drive_test.txt'), 'w') as f:
        f.write('MLFlow persistence test: OK')
        
    # Escribir archivo de prueba en artifacts
    with open(os.path.join(ARTIFACTS_DIR, 'model_test.txt'), 'w') as f:
        f.write('Artifacts persistence test: OK')
        
    print("✅ Todas las carpetas fueron creadas y los archivos de prueba escritos.")
except Exception as e:
    print(f"❌ Error durante el testeo: {e}")
```

```python
# 4. Verificación Final (Listado de Archivos)
print("\n--- Contenido de Drive ---")
print(f"MLRUNS: {os.listdir(MLRUNS_DIR)}")
print(f"ARTIFACTS: {os.listdir(ARTIFACTS_DIR)}")
print("--------------------------")
print("Si ves 'drive_test.txt' y 'model_test.txt', la conexión es CORRECTA.")
```
